## Cleaning gas_data

### 1. Analysing
This dataset is in a format that repeats a lot of extra information in every row (things like codes, labels, and descriptions that are just explaining the same value over and over). Most of these columns don't add anything useful for analysis. To figure out which ones can be removed, we'll count how many different values each column actually contains

```python
print(gas_data.nunique().sort_values())
```

### Identifying useful columns
The dataset has 41 columns, but 37 of them hold the same single value on every row

Only 4 columns contain data that changes from row to row:
- REF_AREA: the country code
- REF_AREA_LABEL: the country name
- TIME_PERIOD: the year
- OBS_VALUE: the export value for that country and year


```python
gas_data.shape
```

```python
gas_data = gas_data[['REF_AREA', 'REF_AREA_LABEL', 'TIME_PERIOD', 'OBS_VALUE']]
gas_data.head(10)
```

### Checking for missing values and duplicates

```python
print(gas_data.isnull().sum())
print(gas_data.duplicated().sum())
```

```python
gas_data['OBS_VALUE'].describe()
```

### Insight

The average export to GDP ratio across all countries and years is about 34%, but the spread is wide. The standard deviation (26.7) is nearly as large as the mean itself, which means export levels vary heavily from one country or year to another rather than clustering close to average

The minimum of 0% likely reflects countries or years with little to no recorded trade activity, or possibly a reporting gap. The maximum of 433.8% stands out. A country cannot literally export more than four times its own GDP through domestic production alone, so this points to a small, trade heavy economy (a re-export or entrepot hub, such as a small island or financial hub economy) where goods pass through the country without being consumed domestically, inflating the ratio

The median (26.8%) is noticeably lower than the mean, which confirms the distribution is right skewed. Most countries sit in a lower, more typical range, while a small number of extreme high exporters pull the average upward

# Servive, Value Data

### Load the Service, Value Data
sva_data = Services, Value Added (NVSRVTOTL.csv)

```python
sva_data = pd.read_csv("Datasets/NVSRVTOTL.csv")
sva_data.shape
```

```python
sva_data.sample(10)
```

### Choosing one consistent measure for sva_data

OBS_VALUE mixes different currencies, price bases, and percentage values, so it can't be used as is. It also has up to 5 versions of the same country-year, one for each report vintage. To get one clean number per country per year, we need to pick one unit, one price basis, and one vintage. We'll use US dollars and constant prices, since USD only exists paired with constant prices in this data. This step checks which vintage has the most complete data for that combination.

```python
sva_data[(sva_data['UNIT_MEASURE'] == 'USD') & (sva_data['PRICE_BASIS'] == 'K')].groupby('VINTAGE').size()
```

### Filtering to one consistent measure

V_SM2026 has the most complete data out of the five vintages, so we filter to US dollars, constant prices, and this vintage, then keep only the columns that matter for analysis: country code, country name, year, and value.

## Cleaning sva_data

This dataset has extra columns that repeat the same information on every row. It also has missing values in OBS_VALUE (133,757 out of 156,946 rows have a value), and columns like PRICE_BASIS, VINTAGE, and UNIT_MEASURE are not constant

```python
sva_data = sva_data[
    (sva_data['UNIT_MEASURE'] == 'USD') &
    (sva_data['PRICE_BASIS'] == 'K') &
    (sva_data['VINTAGE'] == 'V_SM2026')
]

sva_data = sva_data[['REF_AREA', 'REF_AREA_LABEL', 'TIME_PERIOD', 'OBS_VALUE']]
sva_data.shape
```

### Handling missing values in sva_data



```python
sva_data.isnull().sum()
```

1,061 out of 7,056 rows have no value recorded for OBS_VALUE, about 15% of the data. We'll drop these rows since guessing values for a country's yearly output would not be an honest representation of the data.

```python
sva_data = sva_data.dropna(subset=['OBS_VALUE'])
sva_data.isnull().sum()
```

```python
sva_data.duplicated().sum()
```

### Descriptive statistics for sva_data

```python
sva_data['OBS_VALUE'].describe()
```

### Insight

These values are in millions of US dollars at constant prices, so this is the actual size of services output for each country and year, not a percentage.

The mean is about 86,261 million, but the standard deviation (495,994) is far larger than the mean. This means a small number of very large economies are pulling the average up, while most countries sit much lower. The median (7,785 million) confirms this: half of all country-years in this dataset have services output below 7.8 billion, while the mean is dragged up to 86.3 billion by a handful of large economies.

The minimum (about 14 million) likely belongs to a small economy. The maximum (about 12,441,019 million, or 12.4 trillion) likely belongs to a major global economy such as the United States or China.

Since this column measures absolute size rather than a ratio, comparing it directly with gas_data's export percentage would not be meaningful without adjusting for the size of each country's economy.
